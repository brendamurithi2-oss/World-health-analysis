 ## Life expectancy trends across African countries
 zooming into Africa specifically to see how life expectancy has evolved country by contry since 1999

```python
#cheking dataset structure
#check column names and preview the data to see how countries are labelled
```

```python
print(dataset.columns.tolist())
dataset.head()
```

```python
#filter to African countries only using country codes (.isin),
#dataset has no region column and includes regional aggregates
```

```python
african_codes = [
    'DZA','AGO','BEN','BWA','BFA','BDI','CPV','CMR','CAF','TCD','COM','COD','COG','CIV','DJI','EGY','GNQ','ERI','SWZ','ETH','GAB','GMB','GHA','GIN','GNB','KEN','LSO','LBR','LBY','MDG','MWI','MLI','MRT','MUS','MAR','MOZ','NAM','NER','NGA','RWA','STP','SEN','SYC','SLE','SOM','ZAF','SSD','SDN', 'TZA','TGO','TUN','UGA','ZMB','ZWE' 
]
africa = dataset[dataset['country_code'].isin(african_codes)]
print(africa['country'].nunique(),"countries")
africa[['country','year','life_expect']].head()
```

```python
import matplotlib.pyplot as plt
#one line per country
plt.figure(figsize=(12,6))
for country, group in africa.groupby('country'):
    plt.plot(group['year'],group['life_expect'],color='grey',alpha=0.4,linewidth=1)

#Africa average on top
avg = africa.groupby('year')['life_expect'].mean()
plt.plot(avg.index, avg.values, color='red', linewidth=3, label='Africa average')

plt.title('Life Expectancy Trends Across African countries')
plt.xlabel('Year')
plt.ylabel('Life expectancy(years)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
             
```

```python
first_year = africa['year'].min()
last_year = africa['year'].max()

Valid = africa.dropna(subset=['life_expect'])
first_year = Valid['year'].min()
last_year = Valid['year'].max()
print(first_year, last_year)

start = Valid[Valid['year'] == first_year].set_index('country')['life_expect']
end =  Valid[Valid['year'] == last_year].set_index('country')['life_expect']

change = (end - start).dropna().sort_values(ascending=False)

print("Top 5 improvers:")
print(change.head(5))
print("\nBottom 5:")
print(change.tail(5))
```

```python
top_bottom = pd.concat([change.head(5), change.tail(5)])

plt.figure(figsize=(10,6))
colors = ['green'] * 5 + ['red'] * 5
plt.barh(top_bottom.index[::-1],top_bottom.values[::-1], color=colors[::])
plt.title('Change in Life Expectancy, 1999 to 2022 (Top 5 vs Bottom 5)')
plt.xlabel('Years gained')
plt.show()
```

```python
#Did the countries that gain the most also spend more time on health
```

```python
avg_spend = Valid.groupby('country')['health_exp'].mean()
compare = pd.DataFrame(
    {'gain':change, 'avg_health_exp': avg_spend}
).reset_index().dropna()
print(compare.sort_values('gain',ascending=False).head(10))
print("\nCorrelation:", compare['gain'].corr(compare['avg_health_exp']))
```

The correletion between life expectancy gain and average health spending is about 0.20, which is weak. The countries with the biggest gains were mostly low spenders, so spending alone does not explain how much life expectancy improved. Other factors such as disease control and starting level are likely to matter.

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.scatter(compare['avg_health_exp'], compare['gain'], alpha=0.7)
plt.xlabel('Average health expenditure')
plt.ylabel('Life expectancy gain (years)')
plt.title('Do bigger spenders gain more life expectancy?')
plt.show()
```

The scatter plot shows no clear relationship between average health spending and life expectancy gain. Several of the largest gains came from countries with low spending, this supports the weak correlation of about 0.20. Health spending alone does not explain how much life expectancy improved

```python
import matplotlib.pyplot as plt
countries = ['Kenya', 'Rwanda', 'Ethiopia', 'Nigeria', 'South Africa', ]
africa = dataset[dataset['country'].isin(countries)]

plt.figure(figsize=(9,5))
for name in countries:
    data = africa[africa['country'] == name]
    plt.plot(data['year'], data['life_expect'], label=name)

plt.xlabel('year')
plt.ylabel('Life expectancy (years)')
plt.title('Life expectancy trends in selected African countries')
plt.legend()
plt.show()
```

CONCLUSION.

Life expectancy rose in most of the selected African countries between 1999 and 2023. Rwanda and Ethiopia improved the most, while Nigeria stayed lowest and Kenya improved slowly, South Africa fell in the early 2000s and again around 2021 which is likely linked to the HIV epidemic and COVID 19 although this dataset can not prove the cause. 
Together with the earlier analysis this suggests that health spending alone does not explain gains in life expectancy. Countries that started with a low base like Rwanda were able to gain a lot and public health programmes are likely to matter as well.
